```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Banking1.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum() #Cheacking null values
```

```python
df.describe()
```

```python
#Filling all the null values
df["Age"].fillna(df["Age"].mean(), inplace=True)
df["Estimated Income"].fillna(df["Estimated Income"].mean(), inplace=True)
df["Occupation"].fillna(df["Occupation"].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

```python
#Cheacking for any inconsistent spaces
df[df["Name"] != df["Name"].str.strip()]["Name"]
```

```python
df["Name"] = df["Name"].str.strip()
df["Nationality"] = df["Nationality"].str.strip()
```

```python
df[df["Name"] != df["Name"].str.strip()]["Name"]
df[df["Nationality"] != df["Nationality"].str.strip()]["Nationality"]
```

```python
# Changing date to date_time datatype
df["Joined Bank"] = pd.to_datetime(df["Joined Bank"], dayfirst=True, errors='coerce')
```

```python
# Changing age into int datatype
df["Age"] = pd.to_numeric(df["Age"], errors='coerce').astype(int)
```

```python
df.info()
```

```python
# Creating income bands using custom bins
bins=[0, 100000, 300000 , float('inf')]
labels=['Low', 'Medium', 'High']

df["Income Band"]= pd.cut(df["Estimated Income"], bins=bins, labels=labels, include_lowest=True)
```

```python
df["Income Band"].value_counts()
```

```python
categorical_cols = df[["Risk Weighting","Nationality","Occupation","Fee Structure","Loyalty Classification","Properties Owned","Income Band"]].columns
for col in categorical_cols:
    display(df[col].value_counts())
```

```python
# Visualizing distribution of numerical columns
numerical_cols = ['Fee Structure','Age', 'Estimated Income', 'Superannuation Savings', 'Credit Card Balance', 'Bank Loans',
                   'Bank Deposits', 'Checking Accounts', 'Saving Accounts', 'Foreign Currency Account', 'Business Lending']

plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols):
    plt.subplot(4, 3, i + 1)
    sns.histplot(df[col], kde=True)
    plt.title(col)
plt.tight_layout()
plt.show()
```

# Insights
  ### 1. Estimated Income is Right Skewed:
      Banks should offer basic savings and small loans for low-income customers, and wealth management for high-income clients.
  ### 2. Age is Fairly Uniform with a spike at 50:
      Middle-aged customers, at peak earning and saving stages, are the ideal target for loans, deposits, and investment products.
  ### 3. All Financial Columns (Loans, Deposits, Savings) are Heavily Right Skewed:
      Small group of customers hold most of the bank's money.These high-value customers need special retention strategies.

```python
# Visualizing the correlatin matrix using heatmap
numerical_cols = ['Age', 'Estimated Income', 'Superannuation Savings', 'Credit Card Balance',
                  'Bank Loans', 'Bank Deposits', 'Checking Accounts', 'Saving Accounts',
                  'Foreign Currency Account', 'Business Lending', 'Properties Owned']


correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(12, 5))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

# Insights
  ### 1. Deposits and Savings Behavior:
      Strong correlation between bank deposits and savings accounts suggests overlapping behaviors 
  ### 2. Properties Owned has Near Zero Correlation with Everything
      Property ownership alone is not a reliable indicator of banking activity
  ### 3. Bank Loans & Business Lending has Moderate Correlation
      When a customer applies for a personal loan, check if they own a business — high chance they need business lending too.
